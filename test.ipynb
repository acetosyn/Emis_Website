```python
from pathlib import Path
from PIL import Image
import pandas as pd

# Root project folder
root_dir = Path.cwd()

# Culture images folder
image_dir = root_dir / "static" / "images" / "culture"

# Supported extensions
extensions = {".jpg", ".jpeg", ".png", ".webp"}

rows = []

for file in sorted(image_dir.iterdir()):
    if file.is_file() and file.suffix.lower() in extensions:
        try:
            with Image.open(file) as img:
                width, height = img.size
                ratio = round(width / height, 2) if height else None
                rows.append({
                    "Name": file.name,
                    "Width": width,
                    "Height": height,
                    "Ratio": ratio,
                    "Orientation": (
                        "Landscape" if width > height
                        else "Portrait" if height > width
                        else "Square"
                    ),
                    "SizeKB": round(file.stat().st_size / 1024, 1)
                })
        except Exception as e:
            rows.append({
                "Name": file.name,
                "Width": None,
                "Height": None,
                "Ratio": None,
                "Orientation": f"Error: {e}",
                "SizeKB": round(file.stat().st_size / 1024, 1)
            })

df = pd.DataFrame(rows)

# Display in notebook
display(df)

# Save CSV in root folder
output_file = root_dir / "image-properties.csv"
df.to_csv(output_file, index=False)

print(f"Saved to: {output_file}")
```
